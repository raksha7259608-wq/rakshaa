```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
cars_df=pd.read_csv("mtcars.xls")
cars_df
plt.scatter(cars_df['mpg'],cars_df['wt'],marker='*')
plt.xlabel("MPG")
plt.ylabel("Weight")
plt.title('Relationship between the Weight of the carand MPG')
plt.show()
trans=cars_df['am'].value_counts()
print(trans)
trans.plot.bar()
plt.xlabel("Transmission")
plt.title('Bar plot to check the Frequency distribution of Transmission typ')
plt.show()
cars_df['mpg'].plot(kind='box')
plt.xlabel('MPG')
plt.title('Box and Whisker Plot of MPG')
plt.show()
min_value = np.min(cars_df['mpg'])
q1=np.percentile(cars_df['mpg'],25)
median=np.median(cars_df['mpg'])
q3=np.percentile(cars_df['mpg'],75)
max_value = np.max(cars_df['mpg'])
print('MinimumValue:',min_value)
print('Q1:',q1)
print('Median:',median)
print('Q3:',q3)
print('Maximum Value:',max_value)

```
